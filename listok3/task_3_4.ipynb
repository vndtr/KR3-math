# Задание 3.4
## Задача о таутохроне (изохронная кривая)

### Условие:
Найти кривую, по которой время скатывания материальной точки под действием силы тяжести до нижней точки **не зависит от начального положения**: (1) аналитическое решение (постановка, интегральное уравнение, решение); (2) статическая картинка; (3) анимация.

---
## 1. Аналитическое решение

### 1.1. Постановка. Интегральное уравнение для времени спуска
Ось $y$ направим вертикально **вверх**, начало координат поместим в **нижнюю точку** кривой. Точка массы $m$ скользит без трения из состояния покоя с высоты $y=h$. Кривую зададим через длину дуги $s$, отсчитываемую от нижней точки, как функцию высоты: $s=s(y)$, и обозначим $\varphi(y)=\dfrac{ds}{dy}>0$.

**Закон сохранения энергии:** $\dfrac{mv^2}{2}+mgy=mgh\ \Rightarrow\ v=\dfrac{ds}{dt}=-\sqrt{2g(h-y)}$ (минус — высота убывает, $s$ уменьшается). Тогда время спуска от высоты $h$ до нижней точки:
$$
T(h)=\int\frac{ds}{|v|}=\int_0^{h}\frac{\varphi(y)\,dy}{\sqrt{2g(h-y)}} .
$$
Требование таутохронности: $T(h)=T_0$ для всех $h>0$. Получаем **интегральное уравнение Абеля**
$$
\boxed{\ \int_0^{h}\frac{\varphi(y)}{\sqrt{h-y}}\,dy=\sqrt{2g}\,T_0\ \equiv\ c\quad\text{для всех }h>0\ }
$$
Неизвестная функция $\varphi(y)$ стоит под знаком интеграла.

### 1.2. Решение уравнения Абеля
Общее уравнение Абеля: $\displaystyle\int_0^h\frac{\varphi(y)}{\sqrt{h-y}}dy=f(h)$. **Приём решения:** умножим обе части на $\dfrac1{\sqrt{z-h}}$ и проинтегрируем по $h$ от $0$ до $z$:
$$
\int_0^z\frac{f(h)}{\sqrt{z-h}}dh=\int_0^z\varphi(y)\underbrace{\int_y^z\frac{dh}{\sqrt{(z-h)(h-y)}}}_{=\pi}dy=\pi\int_0^z\varphi(y)\,dy
$$
(порядок интегрирования поменяли; внутренний интеграл равен $\pi$ — подстановка $h=y+(z-y)\sin^2\psi$ даёт $\int_0^{\pi/2}2\,d\psi=\pi$). Дифференцируем по $z$:
$$
\varphi(z)=\frac1\pi\frac{d}{dz}\int_0^z\frac{f(h)}{\sqrt{z-h}}\,dh .
$$
Для нашего случая $f=c=\text{const}$: $\displaystyle\int_0^z\frac{c\,dh}{\sqrt{z-h}}=2c\sqrt z$, поэтому
$$
\varphi(y)=\frac{ds}{dy}=\frac{c}{\pi\sqrt y}.
$$

### 1.3. От $s(y)$ к уравнению кривой
Так как $ds^2=dx^2+dy^2$, то $\Bigl(\dfrac{ds}{dy}\Bigr)^2=1+\Bigl(\dfrac{dx}{dy}\Bigr)^2=\dfrac{c^2}{\pi^2y}$. Введём константу $2R:=\dfrac{c^2}{\pi^2}$, т. е. $R=\dfrac{c^2}{2\pi^2}=\dfrac{gT_0^2}{\pi^2}$. Тогда
$$
\Bigl(\frac{dx}{dy}\Bigr)^2=\frac{2R}{y}-1=\frac{2R-y}{y}\ \Longrightarrow\ \frac{dx}{dy}=\sqrt{\frac{2R-y}{y}}\quad(y\le2R).
$$
**Подстановка** $y=R(1-\cos\theta)$, $dy=R\sin\theta\,d\theta=2R\sin\frac\theta2\cos\frac\theta2\,d\theta$. Тогда $\dfrac{2R-y}{y}=\dfrac{1+\cos\theta}{1-\cos\theta}=\cot^2\dfrac\theta2$ и
$$
dx=\cot\frac\theta2\cdot2R\sin\frac\theta2\cos\frac\theta2\,d\theta=2R\cos^2\frac\theta2\,d\theta=R(1+\cos\theta)\,d\theta\ \Longrightarrow\ x=R(\theta+\sin\theta).
$$
(Постоянная интегрирования выбрана так, чтобы $x=0$ в нижней точке.) Итак:
$$
\boxed{\ x=R(\theta+\sin\theta),\qquad y=R(1-\cos\theta),\qquad \theta\in[-\pi,\pi]\ }
$$
Это **циклоида**: кривая, которую описывает точка окружности радиуса $R$, катящейся без проскальзывания по прямой $y=2R$ **снизу** (центр окружности в точке $(R\theta,\,R)$, точка в нижнем положении при $\theta=0$; при $\theta=\pm\pi$ точка касается прямой — острия циклоиды). Связь с временем: $R=\dfrac{gT_0^2}{\pi^2}$, т. е.
$$
T_0=\pi\sqrt{\frac Rg}.
$$

### 1.4. Доказательство независимости времени от начальной точки
**Способ 1 (прямое вычисление).** Для циклоиды $ds=2R\cos\frac\theta2\,d\theta$ (так как $ds^2=R^2[(1+\cos\theta)^2+\sin^2\theta]d\theta^2=4R^2\cos^2\frac\theta2\,d\theta^2$). Начало с $\theta=\theta_0>0$ из состояния покоя: $h-y=R(\cos\theta-\cos\theta_0)=2R\bigl(\sin^2\frac{\theta_0}{2}-\sin^2\frac\theta2\bigr)$. Время:
$$
T=\int_0^{\theta_0}\frac{2R\cos\frac\theta2\,d\theta}{\sqrt{4gR\bigl(\sin^2\frac{\theta_0}2-\sin^2\frac\theta2\bigr)}}=\sqrt{\frac Rg}\int_0^{u_0}\frac{2\,du}{\sqrt{u_0^2-u^2}}=\sqrt{\frac Rg}\cdot2\cdot\frac\pi2=\pi\sqrt{\frac Rg},
$$
где подстановка $u=\sin\frac\theta2$, $du=\frac12\cos\frac\theta2\,d\theta$, $u_0=\sin\frac{\theta_0}2$. Результат **не зависит от $\theta_0$** (т. е. от начальной высоты).

**Способ 2 (через длину дуги — гармонический осциллятор).** Длина дуги от дна: $s=\int_0^\theta2R\cos\frac{\theta'}{2}d\theta'=4R\sin\frac\theta2$, а высота $y=R(1-\cos\theta)=2R\sin^2\frac\theta2=\dfrac{s^2}{8R}$. Уравнение движения вдоль кривой: $\ddot s=-g\dfrac{dy}{ds}=-\dfrac{g}{4R}\,s$ — **уравнение гармонического осциллятора** с частотой $\omega=\sqrt{g/(4R)}$, период которого $4\pi\sqrt{R/g}$ не зависит от амплитуды. Время спуска до дна — четверть периода: $\dfrac{\pi}{2\omega}=\pi\sqrt{R/g}$ — то же самое. Закон движения: $s(t)=s_0\cos\omega t$.

**Замечания.** (1) Для круговой дуги (маятник) $y\approx s^2/(2L)$ лишь приближённо, поэтому время зависит от амплитуды — таутохронной является именно циклоида (Гюйгенс). (2) В нижней точке радиус кривизны циклоиды равен $4R$.

## 2. Код: функции и проверка

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, PillowWriter
from matplotlib.patches import Circle
from scipy.integrate import solve_ivp
from scipy.special import ellipk
from IPython.display import HTML, Image, display
```

```python
g, R = 9.81, 1.0                                     # g, м/с²;  R — радиус образующей окружности, м
T0 = np.pi*np.sqrt(R/g)                              # теоретическое время спуска T0 = π√(R/g)
omega = np.sqrt(g/(4*R))                             # частота колебаний вдоль кривой ω = √(g/(4R))

def cycloid(theta):
    # Циклоида x = R(θ + sin θ), y = R(1 − cos θ), θ ∈ [−π, π]; дно в (0, 0), острия на высоте 2R
    return R*(theta + np.sin(theta)), R*(1 - np.cos(theta))

def theta_of_s(s):
    # Из s = 4R·sin(θ/2) находим угол θ по длине дуги s
    return 2*np.arcsin(np.clip(s/(4*R), -1, 1))

def ball(t, theta0):
    # Положение шарика, начавшего без скорости в точке θ0: s(t) = s0·cos(ωt), s0 = 4R sin(θ0/2)
    s = 4*R*np.sin(theta0/2)*np.cos(omega*t)
    return cycloid(theta_of_s(s))

print(f'T0 = π√(R/g) = {T0:.6f} с;   период колебаний 4·T0 = {4*T0:.6f} с;   ω = {omega:.5f} 1/с')
```

### Проверка 1 (`sympy`): интеграл Абеля и решение $\varphi=c/(\pi\sqrt y)$
Проверяем (а) что $\int_0^h\frac{dy}{\sqrt y\sqrt{h-y}}=\pi$, т. е. решение $\varphi=\frac{c}{\pi\sqrt y}$ удовлетворяет уравнению Абеля; (б) что параметрическая кривая даёт именно $ds/dy=\sqrt{2R/y}$ (то есть $c/\pi=\sqrt{2R}$); (в) время спуска по формуле способа 1.

```python
import sympy as sp

y, h, th, th0, Rs, gs = sp.symbols('y h theta theta0 R g', positive=True)
# (а) интеграл Абеля: подставляем φ(y) = c/(π√y), c сократится
abel = sp.integrate(1/(sp.sqrt(y)*sp.sqrt(h - y)), (y, 0, h))
print('(а) ∫0^h dy/(√y √(h−y)) =', sp.simplify(abel))                 # ожидаем π

# (б) ds/dy для параметрической циклоиды:  ds/dθ = √(x'² + y'²),  dy/dθ = R sin θ
xt = Rs*(th + sp.sin(th)); yt = Rs*(1 - sp.cos(th))
ds_dth = sp.simplify(sp.sqrt(sp.simplify(sp.diff(xt, th)**2 + sp.diff(yt, th)**2)))
print('    ds/dθ =', sp.simplify(sp.trigsimp(ds_dth)))                  # ожидаем 2R|cos(θ/2)|
# сравнение квадратов (чтобы не возиться со знаком корня): (ds/dy)² и 2R/y
lhs = sp.simplify((sp.diff(xt, th)**2 + sp.diff(yt, th)**2)/sp.diff(yt, th)**2)
rhs = 2*Rs/yt
print('(б) (ds/dy)² − 2R/y =', sp.simplify(sp.trigsimp(lhs - rhs)))    # ожидаем 0

# (в) время спуска (способ 1): подстановка u = sin(θ/2) сводит интеграл к ∫ 2du/√(u0² − u²)
u, u0 = sp.symbols('u u0', positive=True)
T_int = sp.sqrt(Rs/gs)*sp.integrate(2/sp.sqrt(u0**2 - u**2), (u, 0, u0))
print('(в) T =', sp.simplify(T_int), '  — не зависит от u0')
```

### Проверка 2 (численная): моделируем движение шарика по циклоиде из уравнений Лагранжа
Независимая проверка: не используем формулу $s=s_0\cos\omega t$, а интегрируем уравнение движения для угла $\theta$. Лагранжиан $L=\frac m2 s'(\theta)^2\dot\theta^2-mgy(\theta)$ даёт $s'^2\ddot\theta+s's''\dot\theta^2=-gy'$, где $s'=2R\cos\frac\theta2$, $s''=-R\sin\frac\theta2$, $y'=R\sin\theta$. Останавливаем счёт в момент, когда $\theta$ проходит $0$ (дно), и измеряем время.

```python
def rhs_cycloid(t, u):
    th, w = u                                          # u = (θ, θ̇)
    sp_, spp, yp = 2*R*np.cos(th/2), -R*np.sin(th/2), R*np.sin(th)
    return [w, (-g*yp - sp_*spp*w**2)/sp_**2]          # θ̈ = (−g y' − s' s'' θ̇²)/s'²

def hit_bottom(t, u): return u[0]                      # событие: θ = 0
hit_bottom.terminal, hit_bottom.direction = True, -1

print(f'теория: T0 = {T0:.8f} с')
for th0 in [0.2, 0.7, 1.4, 2.1, 2.8]:                  # разные начальные точки (чем больше θ0, тем выше старт)
    sol = solve_ivp(rhs_cycloid, (0, 5), [th0, 0.0], events=hit_bottom, rtol=1e-12, atol=1e-13)
    h0 = R*(1 - np.cos(th0))
    print(f'θ0 = {th0:3.1f}  (высота старта {h0:5.3f} м):  T = {sol.t_events[0][0]:.8f} с   отклонение от T0: {sol.t_events[0][0] - T0:+.2e}')
```

### Проверка 3: для круговой дуги (маятника) время зависит от высоты
Сравним с дугой окружности радиуса $L=4R$ (она касается циклоиды в нижней точке с тем же радиусом кривизны). Время спуска из точки с высотой $h$ по окружности: $T_{\rm окр}=\sqrt{L/g}\,K(k)$, $k=\sin(\alpha_0/2)$, $\cos\alpha_0=1-h/L$ ($K$ — полный эллиптический интеграл 1-го рода). При $h\to0$ оба времени совпадают, при росте $h$ время по окружности растёт.

```python
Lc = 4*R                                               # радиус круговой дуги
def T_circle(h):
    alpha0 = np.arccos(1 - h/Lc)                       # угловая амплитуда
    return np.sqrt(Lc/g)*ellipk(np.sin(alpha0/2)**2)   # ellipk принимает параметр m = k²

# проверка формулы численным интегрированием уравнения маятника α̈ = −(g/L) sin α (до α = 0)
def rhs_pend(t, u): return [u[1], -(g/Lc)*np.sin(u[0])]
ev = lambda t, u: u[0]; ev.terminal, ev.direction = True, -1
h_test = 1.5*R; a0 = np.arccos(1 - h_test/Lc)
sol = solve_ivp(rhs_pend, (0, 10), [a0, 0], events=ev, rtol=1e-12, atol=1e-13)
print(f'окружность, h = {h_test} м: численно T = {sol.t_events[0][0]:.8f} с;  формула T = {T_circle(h_test):.8f} с;  циклоида: {T0:.8f} с')
for h_ in [0.05, 0.5, 1.0, 1.5, 2.0]:
    print(f'h = {h_:4.2f} м:   циклоида T = {T0:.5f} с,   окружность T = {T_circle(h_):.5f} с   (отличие {100*(T_circle(h_)/T0 - 1):5.2f} %)')
```

## 3. Статическая визуализация (сохраняется в `task_3_4_static.png`)
**(1)** геометрическое построение циклоиды: окружность радиуса $R$ катится по прямой $y=2R$ снизу, отмеченная точка вычерчивает циклоиду; **(2)** шарики стартуют с разных высот (красные квадраты); точки показывают положения через равные промежутки времени $T_0/6$ — к моменту $T_0$ все шарики оказываются в нижней точке; **(3)** высота шарика от времени: все кривые приходят в $y=0$ в один момент $T_0$; **(4)** время спуска от высоты: циклоида — константа, круговая дуга — растёт (точки — численное решение).

```python
fig, axs = plt.subplots(2, 2, figsize=(15, 10.5))
th_all = np.linspace(-np.pi, np.pi, 600); xc, yc = cycloid(th_all)

# ---- (1) геометрическое построение ----
ax = axs[0, 0]
ax.plot(xc, yc, 'b-', lw=2.5, label='циклоида $x=R(\\theta+\\sin\\theta),\\ y=R(1-\\cos\\theta)$')
ax.axhline(2*R, color='k', lw=1.5); ax.text(-3.2, 2*R + 0.06, 'прямая $y=2R$ (по ней катится окружность)')
for th_, col in zip([-2.4, -1.0, 0.6, 2.0], ['tab:red', 'tab:orange', 'tab:green', 'tab:purple']):
    cx = R*th_                                                              # центр окружности (Rθ, R)
    ax.add_patch(Circle((cx, R), R, fill=False, ec=col, lw=1.2))
    px, py = cycloid(th_)
    ax.plot([cx, px], [R, py], '-', color=col, lw=1); ax.plot(px, py, 'o', color=col, ms=7); ax.plot(cx, R, '.', color=col)
ax.set_xlim(-3.6, 3.6); ax.set_ylim(-0.3, 2.3); ax.set_aspect('equal'); ax.grid(alpha=0.3)
ax.set_xlabel('x, м'); ax.set_ylabel('y, м'); ax.set_title('Геометрическое построение: окружность катится снизу по прямой $y=2R$'); ax.legend(loc='lower center', fontsize=8)

# ---- (2) шарики с разных высот ----
ax = axs[0, 1]; ax.plot(xc, yc, 'b-', lw=2)
starts = [0.5, 1.2, 1.9, 2.5, 3.0]; cols = plt.cm.viridis(np.linspace(0, 0.9, len(starts)))
tk = np.arange(0, 7)*T0/6                                                   # моменты t_k = k·T0/6
for th0, col in zip(starts, cols):
    x0_, y0_ = cycloid(th0); ax.plot(x0_, y0_, 's', color='red', ms=8, zorder=5)          # старт
    bx, by = ball(tk, th0); ax.plot(bx, by, 'o-', color=col, ms=5, lw=0.8, label=f'старт $\\theta_0={th0}$, $h={y0_:.2f}$ м')
ax.plot([], [], 's', color='red', label='начальные положения')
ax.set_xlim(-3.6, 3.6); ax.set_ylim(-0.1, 2.15); ax.set_aspect('equal'); ax.grid(alpha=0.3)
ax.set_xlabel('x, м'); ax.set_ylabel('y, м'); ax.set_title('Спуск с разных высот: положения через $T_0/6$'); ax.legend(fontsize=7.5, loc='upper left')

# ---- (3) высота от времени ----
ax = axs[1, 0]; tt = np.linspace(0, 2*T0, 400)
for th0, col in zip(starts, cols):
    ax.plot(tt, ball(tt, th0)[1], color=col, lw=1.8, label=f'$\\theta_0={th0}$')
ax.axvline(T0, color='k', ls='--'); ax.text(T0 + 0.01, 1.9, '$t=T_0=\\pi\\sqrt{R/g}$')
ax.set_xlabel('t, с'); ax.set_ylabel('высота y, м'); ax.grid(alpha=0.3); ax.legend(fontsize=8)
ax.set_title('Все шарики достигают дна одновременно')

# ---- (4) T(h): циклоида против окружности ----
ax = axs[1, 1]; hh = np.linspace(0.01, 2*R*0.999, 200)
ax.plot(hh, np.full_like(hh, T0), 'b-', lw=2.5, label='циклоида: $T=\\pi\\sqrt{R/g}$ (постоянно)')
ax.plot(hh, [T_circle(x_) for x_ in hh], 'r-', lw=2, label='дуга окружности радиуса $4R$')
num_h = [0.1, 0.5, 1.0, 1.5, 1.9]
num_T = [solve_ivp(rhs_cycloid, (0, 5), [np.arccos(1 - h_/R), 0.0], events=hit_bottom, rtol=1e-11, atol=1e-12).t_events[0][0] for h_ in num_h]
ax.plot(num_h, num_T, 'ko', label='численное решение для циклоиды')
ax.set_ylim(0.9, 1.25); ax.set_xlabel('начальная высота h, м'); ax.set_ylabel('время спуска T, с'); ax.grid(alpha=0.3); ax.legend()
ax.set_title('Таутохронность: циклоида и круговая дуга')
plt.tight_layout()
plt.savefig('task_3_4_static.png', dpi=130, bbox_inches='tight')
plt.show()
```

## 4. Анимация (`FuncAnimation`)
**Слева:** одновременно скатываются пять шариков, стартующих с разных высот; на циклоиде отмечены стартовые точки. Таймер и подписи показывают, что в момент $t=T_0=\pi\sqrt{R/g}$ **все шарики одновременно проходят нижнюю точку** (зелёная вспышка), после чего поднимаются по другой ветви (всего показан полупериод $2T_0$). **Справа:** геометрическое построение циклоиды — окружность радиуса $R$ катится по прямой $y=2R$ снизу, отмеченная точка (красная) вычерчивает кривую; радиус показывает поворот окружности. Внизу справа бегущий маркер на графике высот.

```python
starts = [0.5, 1.2, 1.9, 2.5, 3.0]; cols = plt.cm.viridis(np.linspace(0, 0.9, len(starts)))
n_frames = 90
t_arr = np.linspace(0, 2*T0, n_frames)                              # время анимации: полупериод колебаний
th_roll = np.linspace(-np.pi, np.pi, n_frames)                      # угол качения образующей окружности

fig = plt.figure(figsize=(13, 6), dpi=60)
gs = fig.add_gridspec(2, 2, width_ratios=[1.15, 1], height_ratios=[2.3, 1], hspace=0.35, wspace=0.22)
axL = fig.add_subplot(gs[:, 0]); axR = fig.add_subplot(gs[0, 1]); axH = fig.add_subplot(gs[1, 1])

# --- левая панель: шарики на циклоиде ---
axL.plot(xc, yc, 'b-', lw=2)
balls = [axL.plot([], [], 'o', color=col, ms=12, mec='k', zorder=5)[0] for col in cols]
for th0 in starts: axL.plot(*cycloid(th0), 's', color='red', ms=6, alpha=0.7)
axL.set_xlim(-3.5, 3.5); axL.set_ylim(-0.25, 2.3); axL.set_aspect('equal'); axL.grid(alpha=0.3)
axL.set_xlabel('x, м'); axL.set_ylabel('y, м'); axL.set_title('Одновременный спуск с разных высот')
txt = axL.text(0.02, 0.97, '', transform=axL.transAxes, va='top', family='monospace', fontsize=9, bbox=dict(boxstyle='round', fc='white', alpha=0.9))

# --- правая верхняя панель: построение циклоиды ---
axR.plot(xc, yc, color='lightgray', lw=1.5); axR.axhline(2*R, color='k', lw=1.5)
circ = Circle((0, R), R, fill=False, ec='tab:orange', lw=2); axR.add_patch(circ)
radius, = axR.plot([], [], '-', color='tab:orange'); mark, = axR.plot([], [], 'ro', ms=8); drawn, = axR.plot([], [], 'r-', lw=2)
axR.set_xlim(-3.6, 3.6); axR.set_ylim(-0.2, 2.3); axR.set_aspect('equal'); axR.grid(alpha=0.3)
axR.set_title('Построение: окружность катится по прямой $y=2R$')

# --- правая нижняя панель: высоты шариков ---
tt = np.linspace(0, 2*T0, 300)
for th0, col in zip(starts, cols): axH.plot(tt, ball(tt, th0)[1], color=col, lw=1.3)
axH.axvline(T0, color='k', ls='--', lw=0.8); vl = axH.axvline(0, color='r')
axH.set_xlabel('t, с'); axH.set_ylabel('y, м'); axH.grid(alpha=0.3); axH.set_title('Высоты шариков: все в нуле при $t=T_0$')

def update(k):
    t = t_arr[k]
    for b_, th0 in zip(balls, starts):
        bx, by = ball(t, th0); b_.set_data([bx], [by])                 # положение шарика в момент t
    at_bottom = abs(t - T0) < (t_arr[1] - t_arr[0])*0.51
    txt.set_text(f't = {t:5.3f} с\nT0 = {T0:5.3f} с' + ('\n>>> ВСЕ ШАРИКИ НА ДНЕ <<<' if at_bottom else ''))
    txt.get_bbox_patch().set_facecolor('palegreen' if at_bottom else 'white')
    th_ = th_roll[k]; cx = R*th_; px, py = cycloid(th_)
    circ.center = (cx, R); radius.set_data([cx, px], [R, py]); mark.set_data([px], [py])
    sel = th_roll[:k + 1]; drawn.set_data(*cycloid(sel))
    vl.set_xdata([t, t])
    return balls

anim = FuncAnimation(fig, update, frames=n_frames, interval=60, blit=False)
anim.save('task_3_4_anim.gif', writer=PillowWriter(fps=18), dpi=60)
display(Image(filename='task_3_4_anim.gif'))
plt.rcParams['animation.frame_format'] = 'jpeg'                       # кадры HTML-плеера в JPEG: файл ноутбука заметно меньше
display(HTML(anim.to_jshtml(fps=18)))
plt.close(fig)
```

## 5. Ответ

Таутохрона — **циклоида**:
$$
x=R(\theta+\sin\theta),\qquad y=R(1-\cos\theta),\qquad \theta\in[-\pi,\pi],
$$
образуемая точкой окружности радиуса $R$, катящейся по прямой $y=2R$ снизу. Время скатывания до нижней точки из любого начального положения (без начальной скорости) одинаково:
$$
T_0=\pi\sqrt{\frac Rg}.
$$
Вывод: уравнение Абеля $\int_0^h\frac{\varphi(y)}{\sqrt{h-y}}dy=\sqrt{2g}T_0$ имеет решение $\varphi=\frac{ds}{dy}=\frac{\sqrt{2g}T_0}{\pi\sqrt y}$. Независимость $T$ от старта доказана прямым вычислением и через гармонический осциллятор $\ddot s=-\frac{g}{4R}s$; численные расчёты и сравнение с дугой окружности подтверждают результат.
