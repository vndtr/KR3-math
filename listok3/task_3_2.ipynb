# Задание 3.2
## Бильярд в прямоугольнике

### Условие:
Изучить движение материальной точки (бильярдного шара) в прямоугольной области с упругими отражениями от границ: (1) аналитическое решение (постановка, метод развёртки, анализ типов траекторий, гамильтонова формулировка); (2) статическая картинка (png); (3) анимация.

---
## 1. Аналитическое решение

### 1.1. Постановка задачи
Прямоугольник $\Omega=[0,a]\times[0,b]$ (стороны $a$ и $b$). Шар — материальная точка массы $m$; внутри $\Omega$ сил нет, поэтому движение **равномерное прямолинейное**:
$$
\ddot x=0,\quad \ddot y=0\qquad\Longrightarrow\qquad x=x_0+v_xt,\ \ y=y_0+v_yt .
$$
Скорость $v=\sqrt{v_x^2+v_y^2}$ и энергия $E=\tfrac12mv^2$ сохраняются.

**Граничные условия (упругое отражение).** При ударе о стену угол падения равен углу отражения: **компонента скорости, нормальная к стене, меняет знак, а касательная остаётся прежней**:
- удар о вертикальные стенки $x=0$ или $x=a$: $(v_x,v_y)\mapsto(-v_x,\ v_y)$;
- удар о горизонтальные стенки $y=0$ или $y=b$: $(v_x,v_y)\mapsto(v_x,\ -v_y)$.

В углу происходят оба отражения сразу: $(v_x,v_y)\mapsto(-v_x,-v_y)$ — шар возвращается по тому же пути (такие траектории — особые, их «мало»).

### 1.2. Метод развёртки
Вместо того чтобы отражать **траекторию** от стенки, отразим **саму область** относительно стенки. Тогда отражённая часть траектории продолжает прямую линию в симметричной копии прямоугольника. Повторяя отражения, «развернём» биллиард на плоскость, покрытую зеркальными копиями $\Omega$ (плитки размера $a\times b$). Траектория шара превращается в **прямую**:
$$
X(t)=x_0+v_xt,\qquad Y(t)=y_0+v_yt .
$$
Чтобы вернуться в исходный прямоугольник, нужно «свернуть» плоскость обратно. Для одной координаты свёртка — «пилообразная» функция (треугольная волна):
$$
x(t)=\operatorname{fold}_a(X(t)),\qquad \operatorname{fold}_L(X)=L-\bigl|\,L-(X\bmod 2L)\,\bigr| .
$$
Проверка: при $X\in[0,L]$ получаем $X$; при $X\in[L,2L]$ получаем $2L-X$ (отражение от стенки $x=L$); далее функция $2L$-периодична. Аналогично $y(t)=\operatorname{fold}_b(Y(t))$.

Число ударов к моменту $t$: о вертикальные стенки $n_x=\lfloor X(t)/a\rfloor$, о горизонтальные $n_y=\lfloor Y(t)/b\rfloor$ (при $x_0\in[0,a)$, $y_0\in[0,b)$, $v_x,v_y>0$).

### 1.3. Эквивалентное движение на торе
Функция $\operatorname{fold}_a$ имеет период $2a$, а $\operatorname{fold}_b$ — период $2b$. Поэтому состояние шара зависит только от $(X\bmod 2a,\ Y\bmod 2b)$, то есть от точки **тора** 
$$
\mathbb T^2=\mathbb R^2/(2a\mathbb Z\times2b\mathbb Z).
$$
Введём угловые координаты $\theta_1=\dfrac{\pi X}{a},\ \theta_2=\dfrac{\pi Y}{b}\pmod{2\pi}$. Тогда
$$
\dot\theta_1=\omega_1=\frac{\pi v_x}{a},\qquad \dot\theta_2=\omega_2=\frac{\pi v_y}{b},
$$
то есть бильярд в прямоугольнике эквивалентен **линейному (прямолинейному) потоку на торе** с частотами $\omega_1,\omega_2$. Сам прямоугольник — «фактор» тора по отражениям $\theta_1\to-\theta_1$, $\theta_2\to-\theta_2$ (четыре точки тора отвечают одной точке бильярда).

### 1.4. Периодичность и эргодичность
Траектория замкнётся, если существует $T>0$ с $\theta_1(T)=\theta_1(0)+2\pi p$, $\theta_2(T)=\theta_2(0)+2\pi q$, где $p,q\in\mathbb Z$, т. е. $\omega_1T=2\pi p$, $\omega_2T=2\pi q$. Исключая $T$:
$$
\boxed{\ \frac{\omega_1}{\omega_2}=\frac{v_x/a}{v_y/b}=\frac pq\in\mathbb Q\ \Longleftrightarrow\ \text{траектория периодична}\ }
$$
Период $T=\dfrac{2\pi p}{\omega_1}=\dfrac{2ap}{v_x}=\dfrac{2bq}{v_y}$ (при несократимой дроби $p/q$). За период шар совершает $2p$ ударов о вертикальные и $2q$ ударов о горизонтальные стенки (всего $2(p+q)$). Такие траектории лежат на конечном числе отрезков и **не заполняют** область.

Если $\omega_1/\omega_2$ **иррационально**, то по теореме Кронекера–Вейля траектория на торе **всюду плотна**, а более того — **равномерно распределена**: доля времени, проводимого в любой области $D\subset\Omega$, равна $\dfrac{\operatorname{area}(D)}{ab}$ (поток **эргодичен** на каждом торе). Траектория не замыкается никогда, но подходит к любой точке прямоугольника сколь угодно близко.

### 1.5. Гамильтонова формулировка
Фазовое пространство: $(x,y,p_x,p_y)$, $\ (x,y)\in\Omega$. Гамильтониан
$$
H=\frac{p_x^2+p_y^2}{2m}+U(x,y),\qquad U=\begin{cases}0,&(x,y)\in\Omega,\\ +\infty,&\text{вне }\Omega\end{cases}
$$
(бесконечно высокая потенциальная «яма» с жёсткими стенками). Уравнения Гамильтона внутри $\Omega$:
$$
\dot x=\frac{\partial H}{\partial p_x}=\frac{p_x}{m},\quad \dot y=\frac{p_y}{m},\quad \dot p_x=-\frac{\partial U}{\partial x}=0,\quad \dot p_y=0 .
$$
На стенке импульс скачком меняет знак нормальной компоненты (бесконечная сила на бесконечно малом промежутке времени).

**Интегралы движения.** Сохраняются **по отдельности** $|p_x|$ и $|p_y|$ (а не только энергия $H$): система **вполне интегрируема** (2 степени свободы, 2 независимых интеграла $I_1=|p_x|$, $I_2=|p_y|$). Совместные поверхности уровня — двумерные торы (по теореме Лиувилля–Арнольда).

**Переменные действие–угол.** $J_1=\dfrac1{2\pi}\oint p_x\,dx=\dfrac{1}{2\pi}\cdot2a|p_x|=\dfrac{a|p_x|}{\pi}$, $\ J_2=\dfrac{b|p_y|}{\pi}$ (путь «туда и обратно» длиной $2a$). Гамильтониан в новых переменных:
$$
H=\frac{\pi^2}{2m}\Bigl(\frac{J_1^2}{a^2}+\frac{J_2^2}{b^2}\Bigr),\qquad
\omega_1=\frac{\partial H}{\partial J_1}=\frac{\pi^2J_1}{ma^2}=\frac{\pi v_x}{a},\quad \omega_2=\frac{\partial H}{\partial J_2}=\frac{\pi v_y}{b},
$$
что совпадает с частотами из п. 1.3. Углы эволюционируют линейно: $\theta_i=\theta_i^0+\omega_it$. **Резонанс** $\omega_1/\omega_2\in\mathbb Q$ — периодические траектории, **нерезонансные** значения — эргодическое движение по тору.

## 2. Код: движение, развёртка и проверка метода

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, PillowWriter
from matplotlib.patches import Rectangle
from fractions import Fraction
from IPython.display import HTML, Image, display

a, b = 3.0, 2.0                      # стороны прямоугольника

def fold(X, L):
    # «Свёртка» развёртки обратно в отрезок [0, L]: треугольная волна периода 2L
    return L - np.abs(L - np.mod(X, 2*L))

def billiard(t, x0, y0, vx, vy):
    # Положение (x, y), текущие скорости и числа ударов (nx, ny) в момент(ы) t — по методу развёртки
    X, Y = x0 + vx*t, y0 + vy*t                                  # движение в развёрнутой плоскости: прямая линия
    x, y = fold(X, a), fold(Y, b)                                # свёртываем обратно в прямоугольник
    sx = np.where(np.mod(X, 2*a) < a, 1.0, -1.0)                 # +1 — движемся вправо, −1 — после нечётного числа отражений
    sy = np.where(np.mod(Y, 2*b) < b, 1.0, -1.0)
    return x, y, vx*sx, vy*sy, np.floor(X/a).astype(int), np.floor(Y/b).astype(int)

def billiard_events(t_end, x0, y0, vx, vy):
    # НЕЗАВИСИМАЯ реализация «в лоб»: летим до ближайшей стенки, отражаем скорость, продолжаем.
    # Возвращает список событий [(t, x, y, vx, vy), ...] — состояния сразу ПОСЛЕ каждого удара.
    t, x, y = 0.0, x0, y0
    ev = [(t, x, y, vx, vy)]
    while t < t_end:
        tx = ((a - x) if vx > 0 else (-x))/vx                    # время до вертикальной стенки
        ty = ((b - y) if vy > 0 else (-y))/vy                    # время до горизонтальной стенки
        dt = min(tx, ty)
        x, y, t = x + vx*dt, y + vy*dt, t + dt                   # летим до стенки
        if abs(tx - dt) < 1e-12: vx = -vx                        # отражение от вертикальной стенки (если ударились)
        if abs(ty - dt) < 1e-12: vy = -vy                        # отражение от горизонтальной
        ev.append((t, x, y, vx, vy))
    return ev

def position_from_events(ev, tq):
    # Положение в моменты tq по списку событий (линейная интерполяция внутри отрезка между ударами)
    te = np.array([e[0] for e in ev]); idx = np.searchsorted(te, tq, side='right') - 1
    out = np.zeros((len(tq), 2))
    for j, (tj, i) in enumerate(zip(tq, idx)):
        t0, x, y, vx, vy = ev[i]; out[j] = (x + vx*(tj - t0), y + vy*(tj - t0))
    return out
```

### Проверка 1: метод развёртки (`fold`) совпадает с прямым моделированием отражений
Сравниваем положение шара, полученное двумя независимыми способами, в 3000 моментов времени.

```python
x0, y0, vx, vy = 0.7, 0.4, 4.5, 2.0
ev = billiard_events(30, x0, y0, vx, vy)                         # метод «в лоб»: события-удары
tq = np.linspace(0, 29, 3000)
xy_ev = position_from_events(ev, tq)                              # положение по событиям
xy_fold = np.column_stack(billiard(tq, x0, y0, vx, vy)[:2])       # положение по формуле развёртки
print('число ударов за 30 с:', len(ev) - 1)
print('max |разность положений| =', np.abs(xy_ev - xy_fold).max())    # ожидаем ~1e-12
print('шар всегда внутри прямоугольника:', bool(np.all((xy_fold[:, 0] >= 0) & (xy_fold[:, 0] <= a) & (xy_fold[:, 1] >= 0) & (xy_fold[:, 1] <= b))))
```

### Проверка 2: условие периодичности $\dfrac{v_x/a}{v_y/b}\in\mathbb Q$
Для двух наборов скоростей найдём отношение частот, определим тип траектории и (для рационального случая) период; проверим возвращение в исходное состояние.

```python
def analyze(vx, vy):
    # Отношение частот ω1/ω2 = (vx/a)/(vy/b); рациональность проверяем приближением дробью с малым знаменателем
    rho = (vx/a)/(vy/b)
    fr = Fraction(rho).limit_denominator(50)
    rational = abs(float(fr) - rho) < 1e-12
    return rho, fr, rational

cases = {'рациональный': (4.5, 2.0), 'иррациональный': (4.5, 2*np.sqrt(2))}
for name, (vx, vy) in cases.items():
    rho, fr, rational = analyze(vx, vy)
    print(f'{name:15s}: vx = {vx:.4f}, vy = {vy:.4f},  ω1/ω2 = {rho:.6f}', end='')
    if rational:
        p, q = fr.numerator, fr.denominator
        T = 2*a*p/vx                                              # период: T = 2ap/vx = 2bq/vy
        s0 = billiard(np.array([0.0]), 0.7, 0.4, vx, vy)
        s1 = billiard(np.array([T]), 0.7, 0.4, vx, vy)
        back = all(np.allclose(u, v) for u, v in zip(s0[:4], s1[:4]))     # вернулись в то же положение с той же скоростью?
        print(f'  = {p}/{q}  -> ПЕРИОДИЧЕН, T = {T:.4f} c (проверка 2bq/vy = {2*b*q/vy:.4f}); ударов за период: {2*p} + {2*q} = {2*(p+q)}; возврат в старт: {back}')
    else:
        print('  -> иррациональное отношение: НЕ периодичен (всюду плотен)')
```

### Проверка 3: эргодичность — доля времени в области равна её доле по площади
Возьмём область $D=[0,a/2]\times[0,b/2]$ (четверть площади, т. е. ожидаем $0{,}25$) и посчитаем долю времени, проведённого в ней шаром за очень длинный промежуток времени. Также посчитаем долю клеток сетки $30\times20$, которых коснулась траектория.

```python
def stats(vx, vy, t_end, n=2_000_000):
    t = np.linspace(0, t_end, n)
    x, y = billiard(t, 0.7, 0.4, vx, vy)[:2]
    in_D = np.mean((x <= a/2) & (y <= b/2))                       # доля времени в D
    cells = np.unique(np.floor(x/a*29.999).astype(int)*100 + np.floor(y/b*19.999).astype(int))
    return in_D, len(cells)/(30*20)

for name, (vx, vy) in cases.items():
    in_D, cov = stats(vx, vy, 4000.0)
    print(f'{name:15s}: доля времени в D = {in_D:.4f} (площадь D / площадь = 0.2500);  посещено клеток сетки 30×20: {100*cov:5.1f} %')
```

## 3. Статическая визуализация (сохраняется в `task_3_2_static.png`)
**Верхний ряд:** периодическая траектория (шар вернётся в исходное состояние), эргодическая траектория (заполняет прямоугольник), метод развёртки (прямая в плоскости из зеркальных плиток и та же траектория в исходном прямоугольнике; цифры — одни и те же удары). **Нижний ряд:** линейный поток на торе (развёртка $\theta_1,\theta_2$) для рационального и иррационального случаев и они же на трёхмерном торе.

```python
x0, y0 = 0.7, 0.4
vx_r, vy_r = cases['рациональный']; vx_i, vy_i = cases['иррациональный']
T_r = 4.0                                                          # период рационального случая

def draw_rect(ax):
    # рамка прямоугольника с подписями осей
    ax.add_patch(Rectangle((0, 0), a, b, fill=False, ec='k', lw=2))
    ax.set_xlim(-0.2, a + 0.2); ax.set_ylim(-0.2, b + 0.2); ax.set_aspect('equal')
    ax.set_xlabel('x'); ax.set_ylabel('y'); ax.grid(alpha=0.3)

def wrap_line(t, vx, vy):
    # Координаты на плоском торе [0,2π)² с NaN в местах «склейки», чтобы линия не рисовалась через весь квадрат.
    # Третий результат m[i] — сколько элементов получившегося массива относится к первым i+1 точкам (нужно для анимации).
    th1 = np.mod(np.pi*(x0 + vx*t)/a, 2*np.pi); th2 = np.mod(np.pi*(y0 + vy*t)/b, 2*np.pi)
    j = np.where((np.abs(np.diff(th1)) > np.pi) | (np.abs(np.diff(th2)) > np.pi))[0] + 1   # индексы, перед которыми вставляем NaN
    m = np.arange(1, len(t) + 1) + np.searchsorted(j, np.arange(len(t)), side='right')
    return np.insert(th1, j, np.nan), np.insert(th2, j, np.nan), m

fig = plt.figure(figsize=(18, 11))
gs = fig.add_gridspec(2, 3, hspace=0.28, wspace=0.22)

# (1) периодическая траектория
ax = fig.add_subplot(gs[0, 0]); draw_rect(ax)
t = np.linspace(0, T_r, 2000); xr, yr = billiard(t, x0, y0, vx_r, vy_r)[:2]
ax.plot(xr, yr, 'b-', lw=1.2, label='траектория за период $T=4$ с')
ax.plot(x0, y0, 'ro', ms=9, label='старт'); ax.annotate('', xy=(x0 + 0.35*vx_r/5, y0 + 0.35*vy_r/5), xytext=(x0, y0), arrowprops=dict(arrowstyle='->', color='r', lw=2))
ax.set_title(r'Периодический случай: $\omega_1/\omega_2=3/2$'); ax.legend(loc='upper right', fontsize=8)

# (2) эргодическая траектория
ax = fig.add_subplot(gs[0, 1]); draw_rect(ax)
t = np.linspace(0, 40, 20000); xi, yi = billiard(t, x0, y0, vx_i, vy_i)[:2]
ax.plot(xi, yi, 'g-', lw=0.4, alpha=0.8, label='траектория за 40 с')
ax.plot(x0, y0, 'ro', ms=9, label='старт')
ax.set_title(r'Эргодический случай: $\omega_1/\omega_2$ иррационально'); ax.legend(loc='upper right', fontsize=8)

# (3) метод развёртки: плитки, прямая и те же удары в исходном прямоугольнике
ax = fig.add_subplot(gs[0, 2])
ev = billiard_events(2.6, x0, y0, vx_r, vy_r)[:7]                  # первые удары
t_end = ev[-1][0]
for i in range(0, 5):
    for j in range(0, 4):                                          # плитки: чётные/нечётные — зеркальные копии
        ax.add_patch(Rectangle((i*a, j*b), a, b, fc='lightsteelblue' if (i + j) % 2 == 0 else 'lavender', ec='gray', lw=0.8, alpha=0.7))
ax.add_patch(Rectangle((0, 0), a, b, fill=False, ec='k', lw=2.5))   # исходный прямоугольник
tt = np.linspace(0, t_end, 300)
ax.plot(x0 + vx_r*tt, y0 + vy_r*tt, 'r-', lw=2, label='прямая в развёртке')
ax.plot(x0, y0, 'ro')
for k, (te, xe, ye, _, _) in enumerate(ev[1:], 1):
    ax.plot(x0 + vx_r*te, y0 + vy_r*te, 'ko', ms=5); ax.annotate(str(k), (x0 + vx_r*te, y0 + vy_r*te), xytext=(4, 4), textcoords='offset points', fontsize=9)
# та же траектория, свёрнутая в исходный прямоугольник (внизу слева, масштаб тот же)
ax.plot(fold(x0 + vx_r*tt, a), fold(y0 + vy_r*tt, b), 'b-', lw=1.6, label='та же траектория в $\\Omega$')
ax.set_xlim(-0.2, 5*a + 0.2); ax.set_ylim(-0.2, 4*b + 0.2); ax.set_aspect('equal'); ax.grid(alpha=0.3)
ax.set_xlabel('X'); ax.set_ylabel('Y'); ax.set_title('Метод развёртки: удары 1…6 (цифры) — пересечения линий сетки'); ax.legend(loc='lower right', fontsize=8)

# (4), (5) плоский тор [0,2π)²
for col, (vx, vy, T, name, color) in enumerate([(vx_r, vy_r, 4.0, 'рациональный (замкнутая линия)', 'b'), (vx_i, vy_i, 60.0, 'иррациональный (плотная обмотка)', 'g')]):
    ax = fig.add_subplot(gs[1, col])
    th1, th2, _ = wrap_line(np.linspace(0, T, 6000), vx, vy)
    ax.plot(th1, th2, color + '-', lw=0.6 if col else 1.3)
    ax.plot(np.mod(np.pi*x0/a, 2*np.pi), np.mod(np.pi*y0/b, 2*np.pi), 'ro')
    ax.set_xlim(0, 2*np.pi); ax.set_ylim(0, 2*np.pi); ax.set_aspect('equal'); ax.grid(alpha=0.3)
    ax.set_xlabel(r'$\theta_1=\pi X/a$'); ax.set_ylabel(r'$\theta_2=\pi Y/b$'); ax.set_title('Плоский тор: ' + name)

# (6) трёхмерный тор
ax = fig.add_subplot(gs[1, 2], projection='3d')
Rt, rt = 2.0, 0.8
u, v = np.meshgrid(np.linspace(0, 2*np.pi, 60), np.linspace(0, 2*np.pi, 30))
ax.plot_surface((Rt + rt*np.cos(v))*np.cos(u), (Rt + rt*np.cos(v))*np.sin(u), rt*np.sin(v), color='lightgray', alpha=0.25, linewidth=0)
def torus_xyz(t, vx, vy):
    th1 = np.pi*(x0 + vx*t)/a; th2 = np.pi*(y0 + vy*t)/b
    return (Rt + rt*np.cos(th2))*np.cos(th1), (Rt + rt*np.cos(th2))*np.sin(th1), rt*np.sin(th2)
ax.plot(*torus_xyz(np.linspace(0, 4.0, 1500), vx_r, vy_r), 'b-', lw=2, label='рациональный')
ax.plot(*torus_xyz(np.linspace(0, 25, 8000), vx_i, vy_i), 'g-', lw=0.4, label='иррациональный')
ax.set_box_aspect((1, 1, 0.4)); ax.set_axis_off(); ax.view_init(elev=35, azim=40)
ax.set_title('Те же траектории на трёхмерном торе'); ax.legend(loc='upper left', fontsize=8)
plt.savefig('task_3_2_static.png', dpi=130, bbox_inches='tight')    # сохраняем рисунок
plt.show()
```

## 4. Анимация (`FuncAnimation`)
В каждой строке (верхняя — **рациональный** случай, нижняя — **иррациональный**) три панели: шар в прямоугольнике со следом; **развёртка** (шар движется по прямой через зеркальные плитки, красной звездой — сам шар, синим — след в $\Omega$); **плоский тор** (линия наматывается на тор; в рациональном случае замыкается, в иррациональном — всюду плотно). Динамически показываются время, положение, скорость, число ударов.

```python
# ---- подготовка траекторий ----
T_end, n_frames, n_fine = 8.0, 80, 2000
tf = np.linspace(0, T_end, n_fine)                                  # мелкая сетка времени для следов
scen = [dict(name='Рациональный (ω1/ω2 = 3/2)', vx=vx_r, vy=vy_r, color='tab:blue'),
        dict(name='Иррациональный', vx=vx_i, vy=vy_i, color='tab:green')]
for s in scen:
    s['xy'] = billiard(tf, x0, y0, s['vx'], s['vy'])                 # положение, скорость, удары вдоль всей траектории
    s['th'] = wrap_line(tf, s['vx'], s['vy'])                        # (th1, th2, m): линия на торе с NaN-разрывами

fig, axs = plt.subplots(2, 3, figsize=(14, 7.6), dpi=55, gridspec_kw={'wspace': 0.25, 'hspace': 0.38})
art = []                                                             # сюда складываем изменяемые артисты
for r, s in enumerate(scen):
    ax1, ax2, ax3 = axs[r]
    # --- панель 1: прямоугольник ---
    draw_rect(ax1); tr1, = ax1.plot([], [], '-', color=s['color'], lw=0.8); ball1, = ax1.plot([], [], 'o', color='red', ms=9)
    txt = ax1.text(0.02, 0.97, '', transform=ax1.transAxes, va='top', fontsize=7.5, family='monospace', bbox=dict(boxstyle='round', fc='white', alpha=0.85))
    ax1.set_title(s['name'], fontsize=10)
    # --- панель 2: развёртка ---
    Xmax, Ymax = x0 + s['vx']*T_end, y0 + s['vy']*T_end
    for i in range(int(Xmax // a) + 1):
        for j in range(int(Ymax // b) + 1):
            ax2.add_patch(Rectangle((i*a, j*b), a, b, fc='lightsteelblue' if (i + j) % 2 == 0 else 'lavender', ec='gray', lw=0.3, alpha=0.6))
    ax2.add_patch(Rectangle((0, 0), a, b, fill=False, ec='k', lw=1.5))
    ln2, = ax2.plot([], [], 'r-', lw=1); star2, = ax2.plot([], [], 'r*', ms=11)
    ax2.set_xlim(0, Xmax + 1); ax2.set_ylim(0, Ymax + 1); ax2.set_aspect('equal'); ax2.set_title('Развёртка: прямая в плоскости', fontsize=10)
    # --- панель 3: плоский тор ---
    tr3, = ax3.plot([], [], '-', color=s['color'], lw=0.7); pt3, = ax3.plot([], [], 'ro', ms=7)
    ax3.set_xlim(0, 2*np.pi); ax3.set_ylim(0, 2*np.pi); ax3.set_aspect('equal'); ax3.grid(alpha=0.3)
    ax3.set_xlabel(r'$\theta_1$'); ax3.set_ylabel(r'$\theta_2$'); ax3.set_title('Плоский тор', fontsize=10)
    art.append((tr1, ball1, txt, ln2, star2, tr3, pt3))

def update(k):
    idx = int(k/(n_frames - 1)*(n_fine - 1))                         # индекс текущего момента на мелкой сетке
    t = tf[idx]
    for s, (tr1, ball1, txt, ln2, star2, tr3, pt3) in zip(scen, art):
        x, y, vxc, vyc, nx, ny = s['xy']
        tr1.set_data(x[:idx + 1], y[:idx + 1]); ball1.set_data([x[idx]], [y[idx]])            # шар и след в прямоугольнике
        X, Y = x0 + s['vx']*tf[:idx + 1], y0 + s['vy']*tf[:idx + 1]
        ln2.set_data(X, Y); star2.set_data([X[-1]], [Y[-1]])                                   # прямая в развёртке
        th1, th2, mm = s['th']
        tr3.set_data(th1[:mm[idx]], th2[:mm[idx]])                                             # хвост линии на торе до текущего момента
        pt3.set_data([np.mod(np.pi*X[-1]/a, 2*np.pi)], [np.mod(np.pi*Y[-1]/b, 2*np.pi)])      # текущая точка на торе
        txt.set_text(f't = {t:4.1f} c\nx = {x[idx]:.2f}  y = {y[idx]:.2f}\nvx = {vxc[idx]:+.2f}  vy = {vyc[idx]:+.2f}\nудары: верт. {nx[idx]}, гор. {ny[idx]}')
    return []

anim = FuncAnimation(fig, update, frames=n_frames, interval=60, blit=False)
anim.save('task_3_2_anim.gif', writer=PillowWriter(fps=16), dpi=55)
display(Image(filename='task_3_2_anim.gif'))
plt.rcParams['animation.frame_format'] = 'jpeg'                       # кадры HTML-плеера в JPEG: файл ноутбука заметно меньше
display(HTML(anim.to_jshtml(fps=16)))
plt.close(fig)
```

## 5. Интерактивное управление (`matplotlib.widgets`)
Функция ниже: кнопки «Пауза/Пуск» и «Сброс», ползунок скорости $v_y$ (меняет отношение частот $\omega_1/\omega_2=\dfrac{v_x/a}{v_y/b}$ и тип траектории — на экране показывается рациональное оно или нет). Запуск: `%matplotlib widget` (или `%matplotlib qt`), затем `ui = interactive_billiard()`.

```python
from matplotlib.widgets import Button, Slider

def interactive_billiard():
    fig, ax = plt.subplots(figsize=(8, 6)); plt.subplots_adjust(bottom=0.28)
    draw_rect(ax); trail, = ax.plot([], [], 'b-', lw=0.6); ball, = ax.plot([], [], 'ro', ms=10)
    txt = ax.text(0.02, 0.97, '', transform=ax.transAxes, va='top', family='monospace', bbox=dict(boxstyle='round', fc='white', alpha=0.85))
    st = {'run': True, 't': 0.0, 'vy': 2.0, 'xs': [], 'ys': []}      # состояние: пауза, время, vy, история точек
    VX = 4.5

    def reset(*_):
        st['t'], st['xs'], st['ys'] = 0.0, [], []; trail.set_data([], [])

    def update(_):
        if st['run']:
            st['t'] += 0.02
            x, y = billiard(np.array([st['t']]), 0.7, 0.4, VX, st['vy'])[:2]
            st['xs'].append(x[0]); st['ys'].append(y[0])
            trail.set_data(st['xs'], st['ys']); ball.set_data(x, y)
            rho = (VX/a)/(st['vy']/b); fr = Fraction(rho).limit_denominator(30)
            kind = f'РАЦИОНАЛЬНО ({fr}): периодическая' if abs(float(fr) - rho) < 1e-9 else 'иррационально: эргодическая'
            txt.set_text(f"t = {st['t']:.1f} c\nω1/ω2 = {rho:.4f}\n{kind}")
        return trail, ball, txt

    b1 = Button(plt.axes([0.12, 0.08, 0.18, 0.07]), 'Пауза / Пуск'); b2 = Button(plt.axes([0.34, 0.08, 0.12, 0.07]), 'Сброс')
    sl = Slider(plt.axes([0.55, 0.10, 0.35, 0.04]), 'vy', 0.5, 5.0, valinit=2.0)
    b1.on_clicked(lambda e: st.update(run=not st['run'])); b2.on_clicked(reset)
    sl.on_changed(lambda val: (st.update(vy=val), reset()))
    ani = FuncAnimation(fig, update, interval=30, cache_frame_data=False)
    return ani, b1, b2, sl                                            # ссылки нужно сохранять

# ui = interactive_billiard()    # <- раскомментируйте после %matplotlib widget
```

## 6. Выводы

- Бильярд в прямоугольнике $a\times b$ эквивалентен прямолинейному потоку на торе $\mathbb R^2/(2a\mathbb Z\times2b\mathbb Z)$ с частотами $\omega_1=\pi v_x/a$, $\omega_2=\pi v_y/b$; положение шара $x=\operatorname{fold}_a(x_0+v_xt)$, $y=\operatorname{fold}_b(y_0+v_yt)$.
- Если $\omega_1/\omega_2=p/q\in\mathbb Q$, траектория **периодична** с периодом $T=2ap/v_x=2bq/v_y$ и $2(p+q)$ ударами за период (в примере $3/2$: $T=4$ с, 10 ударов). 
- Если отношение иррационально, траектория **всюду плотна и равномерно распределена** (эргодична): числовая проверка дала долю времени в области $D$, равную $0{,}2499\approx0{,}25$, и покрытие 100 % клеток сетки $30\times20$. (Для рационального примера доля времени в этой области тоже получилась $0{,}25$ — из-за симметрии выбранного $D$, но траектория покрывает лишь около 39 % клеток: различие видно по покрытию, а не по этому показателю.)
- Гамильтонова система вполне интегрируема: сохраняются $|p_x|$ и $|p_y|$, фазовое пространство расслоено на инвариантные торы, на которых движение квазипериодическое.
