# Задание 3.5
## Движение математического маятника

### Условие:
Исследовать движение математического маятника — точечной массы на невесомом нерастяжимом стержне в поле тяжести, включая линейные и нелинейные режимы: (1) вывод уравнения движения, линеаризация, точное решение через эллиптические интегралы, фазовый портрет; (2) статические картинки; (3) анимация.

---
## 1. Аналитическое решение

### 1.1. Вывод уравнения движения
Точечная масса $m$ на невесомом нерастяжимом стержне длины $\ell$ движется по дуге окружности. Угол $\theta$ отсчитываем от нижнего (устойчивого) вертикального положения. На массу действуют сила тяжести $m\vec g$ и сила натяжения стержня $\vec T$ (вдоль стержня). Разложим силы на **касательную** (вдоль движения) и **радиальную** (вдоль стержня) составляющие.

Второй закон Ньютона $m\vec a=m\vec g+\vec T$. Ускорение при движении по окружности: касательное $a_\tau=\ell\ddot\theta$, центростремительное $a_r=\ell\dot\theta^2$ (к центру).
- **Касательное направление.** $T$ перпендикулярна движению и вклада не даёт; тангенциальная составляющая силы тяжести направлена к положению равновесия: $m\ell\ddot\theta=-mg\sin\theta$.
- **Радиальное направление.** $m\ell\dot\theta^2=T-mg\cos\theta$, откуда $T=m(\ell\dot\theta^2+g\cos\theta)$ — натяжение стержня (всегда определяется после решения).

Итак, **уравнение движения математического маятника**:
$$
\boxed{\ \ddot\theta+\omega_0^2\sin\theta=0,\qquad \omega_0^2=\frac g\ell\ }
$$
Оно нелинейно ($\sin\theta$) и не решается в элементарных функциях.

### 1.2. Линеаризация (малые углы)
При $|\theta|\ll1$: $\sin\theta\approx\theta-\theta^3/6+\dots\approx\theta$. Получаем уравнение гармонического осциллятора
$$
\ddot\theta+\omega_0^2\theta=0\ \Longrightarrow\ \theta(t)=\theta_0\cos\omega_0t\quad(\dot\theta(0)=0),
$$
$$
\omega_0=\sqrt{g/\ell},\qquad T_0=2\pi\sqrt{\ell/g}\quad\text{— период не зависит от амплитуды (изохронность).}
$$

### 1.3. Закон сохранения энергии и фазовый портрет
Умножим уравнение на $\dot\theta$ и проинтегрируем: энергия
$$
E=\frac12m\ell^2\dot\theta^2-mg\ell\cos\theta=\text{const}.
$$
Введём безразмерные величины: $y=\dot\theta/\omega_0$, $\ \varepsilon=\dfrac{E}{mg\ell}$. Линии уровня энергии (фазовые траектории): 
$$
\frac{y^2}{2}-\cos\theta=\varepsilon\quad\Longleftrightarrow\quad y=\pm\sqrt{2(\varepsilon+\cos\theta)} .
$$
**Положения равновесия** ($\dot\theta=0$, $\sin\theta=0$): $\theta=0$ и $\theta=\pm\pi$. Линеаризация системы $\dot\theta=\omega_0y$, $\dot y=-\omega_0\sin\theta$ в окрестности точки: матрица $\begin{pmatrix}0&\omega_0\\-\omega_0\cos\theta^*&0\end{pmatrix}$ имеет собственные значения $\lambda^2=-\omega_0^2\cos\theta^*$:
- $\theta^*=0$: $\lambda=\pm i\omega_0$ — **центр** (устойчивое равновесие, малые колебания);
- $\theta^*=\pm\pi$: $\lambda=\pm\omega_0$ — **седло** (неустойчивое верхнее положение).

**Режимы движения** по энергии:
- $-1<\varepsilon<1$ — **колебания** (либрация): замкнутые траектории вокруг центра, угол ограничен амплитудой $\theta_0$, $\varepsilon=-\cos\theta_0$;
- $\varepsilon=1$ — **сепаратриса**: $y=\pm2\cos\frac\theta2$, траектория, идущая в седло (бесконечный период; $\theta(t)=4\arctan e^{\omega_0t}-\pi$);
- $\varepsilon>1$ — **вращение**: маятник делает полные обороты, траектории — волнистые линии, не замкнутые по $\theta$ (но периодические по $\theta\to\theta+2\pi$).

### 1.4. Точное решение через эллиптические интегралы
Для колебаний с амплитудой $\theta_0$ из закона сохранения энергии $\dot\theta^2=2\omega_0^2(\cos\theta-\cos\theta_0)$. Используем $\cos\theta=1-2\sin^2\frac\theta2$ и обозначим $k=\sin\dfrac{\theta_0}{2}$:
$$
\dot\theta^2=4\omega_0^2\Bigl(k^2-\sin^2\frac\theta2\Bigr).
$$
**Подстановка** $\sin\dfrac\theta2=k\sin\varphi$: тогда $\dfrac12\cos\dfrac\theta2\,d\theta=k\cos\varphi\,d\varphi$, $\ \cos\dfrac\theta2=\sqrt{1-k^2\sin^2\varphi}$, а $\dot\theta=2\omega_0k\cos\varphi$. Получаем $dt=\dfrac{d\theta}{\dot\theta}=\dfrac{d\varphi}{\omega_0\sqrt{1-k^2\sin^2\varphi}}$. Четверть периода — от $\theta=0$ ($\varphi=0$) до $\theta_0$ ($\varphi=\pi/2$):
$$
\boxed{\ T=\frac4{\omega_0}\,K(k),\qquad K(k)=\int_0^{\pi/2}\frac{d\varphi}{\sqrt{1-k^2\sin^2\varphi}},\qquad k=\sin\frac{\theta_0}{2}\ }
$$
($K$ — полный эллиптический интеграл 1-го рода). Разложение: 
$$
\frac{T}{T_0}=\frac{2}{\pi}K(k)=1+\frac{\theta_0^2}{16}+\frac{11\,\theta_0^4}{3072}+\dots
$$
Период **растёт** с амплитудой; при $\theta_0\to\pi$ ($k\to1$) $K\to\infty$ (логарифмически) — период стремится к бесконечности (сепаратриса).

**Зависимость $\theta(t)$ (эллиптические функции Якоби).** Обращая $t(\varphi)=\frac1{\omega_0}F(\varphi,k)$ ($F$ — неполный эллиптический интеграл), получаем $\sin\varphi=\operatorname{sn}(\omega_0t,k)$. Если в $t=0$ маятник в крайнем положении (с нулевой скоростью), то
$$
\boxed{\ \theta(t)=2\arcsin\bigl[k\,\operatorname{sn}(\omega_0t+K(k),\,k)\bigr]\ }
$$
(сдвиг на $K$ даёт $\operatorname{sn}K=1$, то есть $\theta(0)=\theta_0$).

**Вращение** ($\varepsilon>1$): время одного оборота $T_{\rm вр}=\dfrac{2\kappa}{\omega_0}K(\kappa)$, $\ \kappa=\sqrt{\dfrac{2}{\varepsilon+1}}<1$ (получается из $T=\int_0^{2\pi}\dfrac{d\theta}{\omega_0\sqrt{2(\varepsilon+\cos\theta)}}$ той же заменой).

## 2. Код: решения и проверка

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, PillowWriter
from scipy.integrate import solve_ivp
from scipy.special import ellipk, ellipj
from IPython.display import HTML, Image, display

g, ell = 9.81, 1.0                              # ускорение свободного падения, длина маятника
w0 = np.sqrt(g/ell); T0 = 2*np.pi/w0            # собственная частота и период малых колебаний
print(f'ω0 = {w0:.5f} 1/с,   T0 = {T0:.5f} с')

def pendulum_rhs(t, u):
    # Система первого порядка: u = (θ, θ̇);  θ̇ = ω,  ω̇ = −ω0² sin θ
    return [u[1], -w0**2*np.sin(u[0])]

def solve_numeric(theta0, omega0, t_eval):
    # Численное решение (метод DOP853 высокого порядка, малые допуски)
    sol = solve_ivp(pendulum_rhs, (t_eval[0], t_eval[-1]), [theta0, omega0], t_eval=t_eval, method='DOP853', rtol=1e-12, atol=1e-13)
    return sol.y[0], sol.y[1]

def theta_linear(t, theta0):
    # линеаризованное решение θ = θ0·cos(ω0 t)
    return theta0*np.cos(w0*t)

def theta_exact(t, theta0):
    # Точное решение через эллиптические функции Якоби: θ = 2 arcsin( k·sn(ω0 t + K, k) ),  k = sin(θ0/2)
    k = np.sin(theta0/2); m = k**2                 # scipy использует параметр m = k²
    sn = ellipj(w0*np.asarray(t) + ellipk(m), m)[0]
    return 2*np.arcsin(k*sn)

def period_exact(theta0):
    # T = (4/ω0)·K(k)
    return 4/w0*ellipk(np.sin(theta0/2)**2)

def period_rotation(eps):
    # Время одного оборота при E/(mgℓ) = eps > 1:  T = 2κ K(κ)/ω0,  κ² = 2/(eps+1)
    m = 2/(eps + 1)
    return 2*np.sqrt(m)*ellipk(m)/w0
```

### Проверка 1: точное решение (через эллиптические функции) совпадает с численным; линейное — только при малых амплитудах
Для нескольких амплитуд сравниваем за 2 периода: $\max|\theta_{\rm числ}-\theta_{\rm точн}|$ и $\max|\theta_{\rm числ}-\theta_{\rm лин}|$; а также относительное изменение энергии (должно быть ≈ 0).

```python
print(' θ0, рад   T/T0 (точн.)  max|числ − точн|   max|числ − лин|   дрейф энергии')
for th0 in [0.05, 0.2, 0.5, 1.0, 2.0, 3.0]:
    t = np.linspace(0, 2*period_exact(th0), 4001)
    th_n, om_n = solve_numeric(th0, 0.0, t)
    E = 0.5*om_n**2 - w0**2*np.cos(th_n)                  # энергия на единицу m ℓ² (E/(mℓ²))
    drift = np.max(np.abs(E - E[0]))/abs(E[0])
    print(f'{th0:5.2f}    {period_exact(th0)/T0:10.6f}     {np.max(np.abs(th_n - theta_exact(t, th0))):12.2e}     {np.max(np.abs(th_n - theta_linear(t, th0))):12.2e}   {drift:10.2e}')
```

### Проверка 2: периоды — измеренные в численном решении и по формулам
Период колебаний измеряем по моментам перехода $\theta$ через 0 снизу вверх (событие `solve_ivp`), для вращения — по времени между переходами $\theta=2\pi k$.

```python
def measured_period_libration(th0):
    ev = lambda t, u: u[0]; ev.direction = 1                       # θ = 0 при движении «вверх» (θ̇ > 0)
    sol = solve_ivp(pendulum_rhs, (0, 4*period_exact(th0)), [0.0, w0*2*np.sin(th0/2)], events=ev, method='DOP853', rtol=1e-12, atol=1e-13)
    return np.diff(sol.t_events[0])[0]                              # интервал между двумя последовательными событиями

def measured_period_rotation(eps):
    om0 = w0*np.sqrt(2*(eps + 1))                                   # скорость в нижней точке (θ=0): y² = 2(ε+1)
    ev = lambda t, u: u[0] - 2*np.pi; ev.direction = 1
    sol = solve_ivp(pendulum_rhs, (0, 30), [0.0, om0], events=ev, method='DOP853', rtol=1e-12, atol=1e-13)
    return sol.t_events[0][0]

print('Колебания:')
for th0 in [0.3, 1.0, 2.0, 3.0]:
    print(f'  θ0 = {th0:3.1f}:  измерено {measured_period_libration(th0):.8f} с;  по формуле 4K(k)/ω0 = {period_exact(th0):.8f} с;  ряд T0(1+θ0²/16+11θ0⁴/3072) = {T0*(1 + th0**2/16 + 11*th0**4/3072):.6f} с')
print('Вращение:')
for eps in [1.2, 2.0, 5.0]:
    print(f'  ε = {eps:3.1f}:  измерено {measured_period_rotation(eps):.8f} с;  по формуле 2κK(κ)/ω0 = {period_rotation(eps):.8f} с')
```

## 4. Анимация (`FuncAnimation`)
**Верхний ряд** — движение маятников в конфигурационном пространстве для трёх амплитуд: $\theta_0=0{,}15$ (малая), $1{,}2$ (умеренная) и $2{,}8$ рад (почти до верхнего положения). Синий маятник — **точное нелинейное** решение, красный (контур) — **линейное** $\theta_0\cos\omega_0t$. При малой амплитуде они совпадают, при умеренной и большой нелинейный маятник «отстаёт» — это переход от гармонических колебаний к нелинейным (период растёт). **Нижний левый график** — те же движения на **фазовой плоскости** $(\theta,\dot\theta/\omega_0)$ (светлые кривые — уровни энергии, красная — сепаратриса), **нижний правый** — $\theta(t)$.
```python
amps = [0.15, 1.2, 2.8]; cols = ['tab:blue', 'tab:green', 'tab:red']
n_frames, t_end = 72, 6.0
t_arr = np.linspace(0, t_end, n_frames)
tf = np.linspace(0, t_end, 1500)                                      # мелкая сетка для графиков и хвостов

def omega_exact(t, th0):
    # θ̇ — численная производная точного решения θ(t) (сетка t должна быть мелкой)
    return np.gradient(theta_exact(t, th0), t)

TH_f = {th0: theta_exact(tf, th0) for th0 in amps}                    # θ(t) на мелкой сетке (считаем один раз)
OM_f = {th0: omega_exact(tf, th0)/w0 for th0 in amps}                 # θ̇/ω0 на мелкой сетке

fig = plt.figure(figsize=(13, 8.2), dpi=50)
gs = fig.add_gridspec(2, 3, height_ratios=[1.15, 1], hspace=0.32, wspace=0.25)
axP = [fig.add_subplot(gs[0, i]) for i in range(3)]
axF = fig.add_subplot(gs[1, 0:2]); axT = fig.add_subplot(gs[1, 2])

# --- верхний ряд: маятники ---
bob_n, rod_n, bob_l, rod_l, txts = [], [], [], [], []
for ax, th0, col in zip(axP, amps, cols):
    ax.set_xlim(-1.25, 1.25); ax.set_ylim(-1.3, 1.25); ax.set_aspect('equal'); ax.axis('off')
    ax.plot([-0.4, 0.4], [0, 0], 'k-', lw=3)                          # потолок
    ax.plot(np.sin(np.linspace(-th0, th0, 60))*ell, -np.cos(np.linspace(-th0, th0, 60))*ell, ':', color='gray', lw=1)   # дуга движения
    rn, = ax.plot([], [], '-', color=col, lw=2.5); bn, = ax.plot([], [], 'o', color=col, ms=16, mec='k')
    rl, = ax.plot([], [], '--', color='crimson', lw=1.2); bl, = ax.plot([], [], 'o', mfc='none', mec='crimson', ms=16, mew=2)
    ax.set_title(f'$\\theta_0={th0}$ рад;  $T/T_0={period_exact(th0)/T0:.3f}$', fontsize=11)
    txts.append(ax.text(0.5, 0.0, '', transform=ax.transAxes, ha='center', va='bottom', fontsize=8.5, family='monospace'))
    bob_n.append(bn); rod_n.append(rn); bob_l.append(bl); rod_l.append(rl)
axP[0].plot([], [], 'o', color='tab:blue', label='точное (нелинейное)'); axP[0].plot([], [], 'o', mfc='none', mec='crimson', label='линейное $\\theta_0\\cos\\omega_0t$')
axP[0].legend(loc='upper left', fontsize=8)

# --- нижний левый: фазовая плоскость ---
TH, Y = np.meshgrid(np.linspace(-np.pi - 0.3, np.pi + 0.3, 500), np.linspace(-3.2, 3.2, 400))
axF.contour(TH, Y, Y**2/2 - np.cos(TH), levels=np.linspace(-0.95, 4, 18), colors='lightgray', linewidths=0.7)
axF.contour(TH, Y, Y**2/2 - np.cos(TH), levels=[1.0], colors='red', linewidths=2)
tails, dots = [], []
for th0, col in zip(amps, cols):
    full_t = np.linspace(0, 2*period_exact(th0), 800)
    axF.plot(theta_exact(full_t, th0), omega_exact(full_t, th0)/w0, '-', color=col, lw=0.6, alpha=0.4)
    tl, = axF.plot([], [], '-', color=col, lw=2); dt_, = axF.plot([], [], 'o', color=col, ms=9, mec='k')
    tails.append(tl); dots.append(dt_)
axF.plot([0], [0], 'bo'); axF.plot([np.pi, -np.pi], [0, 0], 'rx', ms=9)
axF.set_xlim(-np.pi - 0.3, np.pi + 0.3); axF.set_ylim(-3.2, 3.2); axF.grid(alpha=0.3)
axF.set_xlabel('θ, рад'); axF.set_ylabel('$\\dot\\theta/\\omega_0$'); axF.set_title('Фазовая плоскость (красная — сепаратриса)')

# --- нижний правый: θ(t) ---
for th0, col in zip(amps, cols): axT.plot(tf, theta_exact(tf, th0), color=col, lw=1.3)
axT.set_xlabel('t, с'); axT.set_ylabel('θ, рад'); axT.grid(alpha=0.3); axT.set_title('θ(t)')
vline = axT.axvline(0, color='k', lw=0.8); dots_t = [axT.plot([], [], 'o', color=col, mec='k')[0] for col in cols]

def update(k):
    t = t_arr[k]; idx = int(k/(n_frames - 1)*(len(tf) - 1))
    for i, th0 in enumerate(amps):
        th_n = float(theta_exact(t, th0)); th_l = float(theta_linear(t, th0))
        rod_n[i].set_data([0, ell*np.sin(th_n)], [0, -ell*np.cos(th_n)]); bob_n[i].set_data([ell*np.sin(th_n)], [-ell*np.cos(th_n)])
        rod_l[i].set_data([0, ell*np.sin(th_l)], [0, -ell*np.cos(th_l)]); bob_l[i].set_data([ell*np.sin(th_l)], [-ell*np.cos(th_l)])
        txts[i].set_text(f't={t:4.2f} c  θ={th_n:+.2f}  Δ(θ_лин−θ)={th_l - th_n:+.2f}')
        tails[i].set_data(TH_f[th0][:idx + 1], OM_f[th0][:idx + 1])           # хвост траектории на фазовой плоскости
        dots[i].set_data([TH_f[th0][idx]], [OM_f[th0][idx]])                  # текущая точка на фазовой плоскости
        dots_t[i].set_data([t], [th_n])
    vline.set_xdata([t, t])
    return []

anim = FuncAnimation(fig, update, frames=n_frames, interval=55, blit=False)
anim.save('task_3_5_anim.gif', writer=PillowWriter(fps=18), dpi=50)
display(Image(filename='task_3_5_anim.gif'))
plt.rcParams['animation.frame_format'] = 'jpeg'                       # кадры HTML-плеера в JPEG: файл ноутбука заметно меньше
display(HTML(anim.to_jshtml(fps=18)))
plt.close(fig)
```

## 5. Выводы и ответ

- Уравнение движения: $\ddot\theta+\dfrac g\ell\sin\theta=0$; энергия $E=\tfrac12m\ell^2\dot\theta^2-mg\ell\cos\theta$ сохраняется.
- **Малые углы:** $\theta=\theta_0\cos\omega_0t$, $T_0=2\pi\sqrt{\ell/g}$ — не зависит от амплитуды.
- **Нелинейный режим:** $T=\dfrac{4}{\omega_0}K\bigl(\sin\frac{\theta_0}2\bigr)=T_0\bigl(1+\frac{\theta_0^2}{16}+\frac{11\theta_0^4}{3072}+\dots\bigr)$ — период растёт с амплитудой и стремится к бесконечности при $\theta_0\to\pi$; точное решение $\theta(t)=2\arcsin[k\,\mathrm{sn}(\omega_0t+K,k)]$ (проверено численным интегрированием с точностью $\sim10^{-12}$).
- **Фазовый портрет:** центр в $\theta=0$, сёдла в $\theta=\pm\pi$, сепаратриса $\dot\theta=\pm2\omega_0\cos\frac\theta2$ ($E=mg\ell$) отделяет колебания ($E<mg\ell$) от вращений ($E>mg\ell$); период вращения $T=\dfrac{2\kappa}{\omega_0}K(\kappa)$, $\kappa=\sqrt{2/(\varepsilon+1)}$.
