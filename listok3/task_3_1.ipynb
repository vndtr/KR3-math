# Задание 3.1
## Задача о равномерном спуске

### Условие:
Найти такую кривую, что тяжёлая точка под действием силы тяжести **равномерно** опускается по этой кривой (кривая расположена в вертикальной плоскости). Требуется: (1) подробное аналитическое решение; (2) статическая картинка (png); (3) анимация решения.

---
## 1. Аналитическое решение

### 1.1. Постановка и система координат
Введём систему координат: ось $x$ — горизонталь, ось $y$ — вертикаль **вверх**. Точка массы $m$ движется по гладкой (без трения) кривой $y=f(x)$. Условие «равномерного опускания» означает, что **вертикальная скорость постоянна**:
$$
v_y=\dot y=-v_0=\text{const}<0,\qquad v_0>0.
$$
Горизонтальная скорость $v_x=\dot x$ при этом, вообще говоря, меняется.

### 1.2. Силы и второй закон Ньютона
На точку действуют две силы:
- сила тяжести $m\vec g=(0,\,-mg)$;
- **сила нормальной реакции** кривой $\vec N$. Кривая гладкая, поэтому $\vec N$ перпендикулярна касательной. Касательный вектор к графику: $(1,f')$, поэтому единичная нормаль, направленная вверх: 
$$\vec n=\frac{(-f',\,1)}{\sqrt{1+f'^2}},\qquad \vec N=N\,\vec n,\quad N>0.$$

Второй закон Ньютона $m\ddot{\vec r}=m\vec g+\vec N$ в проекциях:
$$
\begin{cases}
m\ddot x=-\dfrac{N f'}{\sqrt{1+f'^2}},\\[3mm]
m\ddot y=-mg+\dfrac{N}{\sqrt{1+f'^2}}.
\end{cases}
$$

### 1.3. Исключение силы нормальной реакции
Так как $\dot y=-v_0=\text{const}$, то $\ddot y=0$. Из второго уравнения 
$$\frac{N}{\sqrt{1+f'^2}}=mg\quad\Longrightarrow\quad N=mg\sqrt{1+f'^2}.$$
(Реакция **больше** веса: кривая должна не только держать точку, но и «разгонять» её по горизонтали.) Подставим это в первое уравнение:
$$
m\ddot x=-mg\,f'(x)\quad\Longrightarrow\quad \boxed{\ \ddot x=-g\,f'(x)\ }.
$$
Масса сократилась: форма кривой от массы не зависит.

### 1.4. Получение дифференциального уравнения для кривой
Условие $\dot y=-v_0$ с учётом $y=f(x)$ даёт $f'(x)\,\dot x=-v_0$, откуда
$$
\dot x=-\frac{v_0}{f'(x)}.
$$
Продифференцируем по времени (цепное правило: $\frac{d}{dt}=\dot x\frac{d}{dx}$):
$$
\ddot x=\frac{v_0 f''}{f'^2}\,\dot x=\frac{v_0f''}{f'^2}\cdot\Bigl(-\frac{v_0}{f'}\Bigr)=-\frac{v_0^2f''}{f'^3}.
$$
Приравниваем к $\ddot x=-gf'$:
$$
-\frac{v_0^2f''}{f'^3}=-gf'\quad\Longrightarrow\quad \boxed{\ v_0^2\,f''=g\,(f')^4\ }.
$$
Это дифференциальное уравнение второго порядка относительно $f(x)$.

### 1.5. Интегрирование
**Первый интеграл.** Обозначим $p=f'(x)$. Уравнение становится уравнением с разделяющимися переменными: $\dfrac{dp}{dx}=\dfrac{g}{v_0^2}p^4$, т. е. $\dfrac{dp}{p^4}=\dfrac{g}{v_0^2}dx$. Интегрируем:
$$
-\frac{1}{3p^3}=\frac{g}{v_0^2}(x-x_0)\ \Longrightarrow\ p^3=-\frac{v_0^2}{3g\,(x-x_0)}\ \Longrightarrow\ 
f'(x)=-\sqrt[3]{\frac{v_0^2}{3g\,(x-x_0)}}.
$$
($x_0$ — произвольная постоянная.) При $x>x_0$ получаем $f'<0$, и из $\dot x=-v_0/f'$ видим $\dot x>0$: точка движется вправо и вниз. При $x<x_0$ знаки обратные: $f'>0$, $\dot x<0$ — точка движется влево и вниз (левая ветвь, симметричная правой).

**Второй интеграл.** Для правой ветви ($x>x_0$):
$$
f(x)=-\Bigl(\frac{v_0^2}{3g}\Bigr)^{1/3}\int (x-x_0)^{-1/3}dx=y_0-\frac32\Bigl(\frac{v_0^2}{3g}\Bigr)^{1/3}(x-x_0)^{2/3},
$$
где $y_0$ — вторая произвольная постоянная. С учётом обеих ветвей:
$$
\boxed{\ y=y_0-\frac32\Bigl(\frac{v_0^2}{3g}\Bigr)^{1/3}|x-x_0|^{2/3}\ },\qquad\text{или}\qquad (y_0-y)^3=\frac{9v_0^2}{8g}\,(x-x_0)^2 .
$$
Это **полукубическая парабола (парабола Нейла)** с точкой возврата (остриём) в $(x_0,y_0)$, направленным вверх.

### 1.6. Закон движения по времени
Из $\dot y=-v_0$: $y(t)=y_0-v_0t$ (отсчёт времени от остриё). Подставляя $y_0-y=v_0t$ в уравнение кривой, найдём
$$
x(t)=x_0\pm\frac{\sqrt{8gv_0}}{3}\,t^{3/2},\qquad 
v_x=\pm\sqrt{2gv_0t},\qquad N=mg\sqrt{1+\frac{v_0^2}{v_x^2}}=mg\sqrt{1+\frac{v_0}{2gt}}.
$$
Проверка: $\ddot x=\pm\sqrt{\dfrac{gv_0}{2t}}$, а $-gf'=g\,\dfrac{v_0}{|v_x|}=\sqrt{\dfrac{gv_0}{2t}}$ — совпадает.

### 1.7. Начальные условия
Пусть в начальный момент точка находится в $(x_s,y_s)$, горизонтальная скорость $v_{xs}\neq0$, вертикальная $-v_0$. Тогда $f'(x_s)=-v_0/v_{xs}$, и из формулы для $f'$ получаем положение острия:
$$
x_0=x_s-\frac{v_{xs}^3}{3gv_0},\qquad y_0=y_s+\frac32\Bigl(\frac{v_0^2}{3g}\Bigr)^{1/3}|x_s-x_0|^{2/3}.
$$
Так для каждого набора $(x_s,y_s,v_{xs},v_0)$ кривая определяется однозначно. Если стартовать ровно из острия ($v_{xs}=0$), это идеализация: там $f'\to\infty$ и $N\to\infty$.

**Замечание.** Кроме найденных кривых, решением является вертикальная прямая, но на ней $N=0$ и точка падает свободно с ускорением $g$ (неравномерно), поэтому она не подходит.

## 2. Код: кривая, начальные условия и проверка решения

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, PillowWriter
from IPython.display import HTML, Image, display

g = 9.81                                    # ускорение свободного падения, м/с²

def c0(v0):
    # коэффициент перед |x − x0|^{2/3}:  (3/2)·(v0²/(3g))^{1/3}
    return 1.5*(v0**2/(3*g))**(1/3)

def curve_y(x, x0, y0, v0):
    # уравнение кривой y(x) (обе ветви сразу, поэтому |x − x0|)
    return y0 - c0(v0)*np.abs(x - x0)**(2/3)

def from_initial(xs, ys, vxs, v0):
    # По начальным условиям (xs, ys, vxs, v0) находит параметры кривой:
    # x0, y0 — координаты острия, sigma = ±1 — ветвь (знак горизонтальной скорости),
    # t_s — время «от острия» до стартовой точки.
    x0 = xs - vxs**3/(3*g*v0)               # формула из п. 1.7
    y0 = ys + c0(v0)*abs(xs - x0)**(2/3)
    return x0, y0, np.sign(vxs), (y0 - ys)/v0

def motion(t, x0, y0, v0, sigma):
    # Закон движения (п. 1.6), t — время, отсчитываемое от острия
    x  = x0 + sigma*np.sqrt(8*g*v0)/3*t**1.5
    y  = y0 - v0*t
    vx = sigma*np.sqrt(2*g*v0*t)
    with np.errstate(divide='ignore'):
        N_over_mg = np.sqrt(1 + v0**2/vx**2)   # N/(mg) = √(1 + v0²/vx²)
    return x, y, vx, -v0*np.ones_like(t), N_over_mg
```

### Проверка 1 (символьная, `sympy`): подставим найденную кривую в уравнение $v_0^2f''=g\,(f')^4$

```python
import sympy as sp

x, v0s, gs, y0s = sp.symbols('x v0 g y0', positive=True)     # x > 0: правая ветвь при x0 = 0
f = y0s - sp.Rational(3, 2)*(v0s**2/(3*gs))**sp.Rational(1, 3)*x**sp.Rational(2, 3)
residual = sp.simplify(v0s**2*sp.diff(f, x, 2) - gs*sp.diff(f, x)**4)
print('v0² f\'\' − g (f\')⁴ =', residual)          # должно быть 0
```

### Проверка 2 (численная): интегрируем уравнение движения $\ddot x=-gf'(x)$ и убеждаемся, что $\dot y=-v_0$ постоянна
Это независимая проверка: мы не используем готовый закон $x(t)$, а решаем ОДУ численно (`scipy.integrate.solve_ivp`) и сравниваем.

```python
from scipy.integrate import solve_ivp

# начальные условия: стартуем в точке (2, 1) с горизонтальной скоростью 1.5 м/с, вертикальная скорость −1 м/с
xs, ys, vxs, v0 = 2.0, 1.0, 1.5, 1.0
x0, y0, sigma, t_s = from_initial(xs, ys, vxs, v0)
print(f'острие кривой: x0 = {x0:.4f}, y0 = {y0:.4f}; время от острия до старта t_s = {t_s:.4f} с')

def fprime(x):
    # f'(x) = −(v0²/(3g(x − x0)))^{1/3}  (для x > x0), для x < x0 — со знаком «+»
    return -np.sign(x - x0)*(v0**2/(3*g*np.abs(x - x0)))**(1/3)

def rhs(t, s):
    # s = (x, vx);  ẋ = vx,  v̇x = −g f'(x)   (уравнение из п. 1.3)
    return [s[1], -g*fprime(s[0])]

T = 1.5                                         # интегрируем 1.5 с
sol = solve_ivp(rhs, (0, T), [xs, vxs], rtol=1e-10, atol=1e-12, dense_output=True)
t = np.linspace(0, T, 7)
xn, vxn = sol.sol(t)
vy = fprime(xn)*vxn                             # вертикальная скорость ẏ = f'(x)·ẋ
print('численная вертикальная скорость ẏ (должна быть −1):', np.round(vy, 8))
xa = motion(t + t_s, x0, y0, v0, sigma)[0]     # аналитическое x(t) из п. 1.6 (время сдвинуто на t_s)
print('max |x_числ − x_аналит| =', np.max(np.abs(xn - xa)))
```

## 3. Статическая визуализация (сохраняется в `task_3_1_static.png`)
На рисунке: (слева) кривые для разных начальных условий — стартовое положение отмечено красным квадратом, остриё — чёрным треугольником, точки на кривой — положения через равные промежутки времени 0,5 с (по вертикали они расположены **равномерно**); (справа) графики $v_x(t)$, $v_y(t)$ и $N/(mg)$ для $v_0=1$ м/с.

```python
# Набор начальных условий: (xs, ys, vxs, v0, подпись, цвет)
cases = [
    (0.0, 0.0,  0.0, 0.5, r'$v_0=0.5$, старт в острие', 'tab:blue'),
    (0.0, 0.0,  0.0, 1.0, r'$v_0=1$, старт в острие',   'tab:green'),
    (0.0, 0.0,  0.0, 2.0, r'$v_0=2$, старт в острие',   'tab:orange'),
    (3.0, -1.0, 1.5, 1.0, r'старт $(3,-1)$, $v_{x0}=1.5$, $v_0=1$', 'tab:purple'),
    (-4.0, 0.5, -2.0, 1.5, r'старт $(-4,0.5)$, $v_{x0}=-2$, $v_0=1.5$', 'tab:brown'),
]
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5.5), gridspec_kw={'width_ratios': [1.5, 1]})
xx = np.linspace(-9, 9, 2000)
for xs, ys, vxs, v0, lab, col in cases:
    if vxs == 0:                                   # старт точно в острие: x0 = xs, y0 = ys
        x0, y0, sigma, t_s = xs, ys, 1.0, 0.0
    else:
        x0, y0, sigma, t_s = from_initial(xs, ys, vxs, v0)
    ax1.plot(xx, curve_y(xx, x0, y0, v0), color=col, lw=1.8, label=lab)             # сама кривая
    ax1.plot(x0, y0, '^', color='k', ms=7)                                          # остриё
    ax1.plot(xs, ys, 's', color='red', ms=9, zorder=5)                              # стартовое положение
    tt = t_s + np.arange(0, 6, 0.5)                                                 # моменты времени через 0.5 с
    px, py = motion(tt, x0, y0, v0, sigma)[:2]
    ax1.plot(px, py, '.', color=col, ms=8)                                          # положения точки
ax1.plot([], [], 's', color='red', label='начальное положение')
ax1.plot([], [], '^', color='k', label='остриё $(x_0,y_0)$')
ax1.set_xlim(-9, 9); ax1.set_ylim(-4, 1.5)
ax1.set_xlabel('x, м'); ax1.set_ylabel('y, м'); ax1.grid(alpha=0.3)
ax1.set_title('Кривые равномерного спуска: полукубические параболы')
ax1.legend(fontsize=8, loc='lower center', ncol=2)

# правый график: зависимость скоростей и реакции от времени (v0 = 1, старт из острия)
t = np.linspace(0.02, 4, 400)
x_, y_, vx_, vy_, N_ = motion(t, 0, 0, 1.0, 1.0)
ax2.plot(t, vx_, label=r'$v_x(t)=\sqrt{2gv_0t}$')
ax2.plot(t, vy_, label=r'$v_y(t)=-v_0$ (постоянна)')
ax2.plot(t, N_, label=r'$N/(mg)$')
ax2.set_xlabel('t, с'); ax2.set_ylabel('скорость, м/с  |  $N/(mg)$'); ax2.set_ylim(-3, 10)
ax2.grid(alpha=0.3); ax2.legend(); ax2.set_title('Параметры движения ($v_0=1$ м/с)')
plt.tight_layout()
plt.savefig('task_3_1_static.png', dpi=150, bbox_inches='tight')      # сохраняем рисунок в png
plt.show()
```

## 4. Анимация (`FuncAnimation`)
Точка равномерно (с постоянной вертикальной скоростью) опускается по кривой. Показаны: след траектории, вектор скорости (зелёный), вектор силы тяжести (серый), сила нормальной реакции (красный) и динамически обновляемые параметры движения. Внизу — графики $y(t)$ и $v_x(t)$, $N/mg$ с бегущим маркером. Анимация показана в двух вариантах: GIF и интерактивная HTML-версия (кнопки ⏮ ⏪ ▶ ⏸ ⏩ ⏭, ползунок скорости, режимы Once/Loop/Reflect).

```python
# ---- параметры анимации ----
v0 = 1.0                              # вертикальная скорость спуска, м/с
x0, y0, sigma = 0.0, 0.0, 1.0         # острие в начале координат, правая ветвь
T_end, n_frames = 2.2, 70             # длительность, с; число кадров
T = np.linspace(0.01, T_end, n_frames)
X, Y, VX, VY, NR = motion(T, x0, y0, v0, sigma)      # заранее считаем всё движение
xx = np.linspace(0, X[-1] + 0.5, 600)

fig = plt.figure(figsize=(10, 6.2), dpi=70)
gs = fig.add_gridspec(2, 2, height_ratios=[1.7, 1], hspace=0.38, wspace=0.28)
ax = fig.add_subplot(gs[0, :]); axa = fig.add_subplot(gs[1, 0]); axb = fig.add_subplot(gs[1, 1])

# --- главная панель: кривая и точка ---
ax.plot(xx, curve_y(xx, x0, y0, v0), 'b-', lw=1.8, label=r'$y=y_0-c\,x^{2/3}$')
ax.plot(x0, y0, 'k^', ms=8, label='остриё (старт)')
trail, = ax.plot([], [], 'r-', lw=1, alpha=0.5)                   # след
point, = ax.plot([], [], 'ro', ms=11, zorder=5, label='точка')
# стрелки (Quiver): скорость, сила тяжести и реакция опоры; angles='xy' — длина в единицах осей
arr_v = ax.quiver([0], [0], [0], [0], angles='xy', scale_units='xy', scale=1, color='green', width=0.004)
arr_g = ax.quiver([0], [0], [0], [0], angles='xy', scale_units='xy', scale=1, color='gray', width=0.004)
arr_N = ax.quiver([0], [0], [0], [0], angles='xy', scale_units='xy', scale=1, color='red', width=0.004)
ax.plot([], [], color='green', label=r'скорость $\vec v$'); ax.plot([], [], color='gray', label=r'$m\vec g$'); ax.plot([], [], color='red', label=r'$\vec N$')
ax.set_xlim(-0.5, X[-1] + 1); ax.set_ylim(Y[-1] - 0.6, 0.6); ax.set_aspect('equal')
ax.set_xlabel('x, м'); ax.set_ylabel('y, м'); ax.grid(alpha=0.3); ax.legend(loc='upper right', fontsize=8, ncol=3)
info = ax.text(0.01, 0.04, '', transform=ax.transAxes, fontsize=9, va='bottom',
               bbox=dict(boxstyle='round', fc='white', alpha=0.85), family='monospace')

# --- нижние панели: графики с бегущим маркером ---
tt = np.linspace(0.01, T_end, 300); _, yy, vxx, vyy, NN = motion(tt, x0, y0, v0, sigma)
axa.plot(tt, yy, 'b-', label='y(t)'); axa.plot(tt, motion(tt, x0, y0, v0, sigma)[0], 'm-', label='x(t)')
axa.set_xlabel('t, с'); axa.set_ylabel('м'); axa.grid(alpha=0.3); axa.legend(fontsize=8)
axb.plot(tt, vxx, 'g-', label='$v_x$'); axb.plot(tt, vyy, 'k--', label='$v_y$'); axb.plot(tt, NN, 'r-', label='$N/mg$')
axb.set_xlabel('t, с'); axb.set_ylim(-1.5, 7.5); axb.grid(alpha=0.3); axb.legend(fontsize=8)
mark_a, = axa.plot([], [], 'ro'); mark_b, = axb.plot([], [], 'ko'); vline_b = axb.axvline(0, color='gray', lw=0.8)

def update(k):
    # Функция кадра k: обновляет положение точки, стрелки, текст и маркеры
    point.set_data([X[k]], [Y[k]]); trail.set_data(X[:k+1], Y[:k+1])
    arr_v.set_offsets([[X[k], Y[k]]]); arr_v.set_UVC([0.5*VX[k]], [0.5*VY[k]])      # скорость (масштаб 0.5 с)
    arr_g.set_offsets([[X[k], Y[k]]]); arr_g.set_UVC([0], [-0.5])                   # сила тяжести — всегда вниз
    fp = -VY[k]/VX[k]*1.0                                                            # |f'| = v0/vx
    nx, ny = (fp, 1)/np.hypot(fp, 1)                                                 # единичная нормаль n = (−f', 1)/√(1+f'²), f' = −fp
    arr_N.set_offsets([[X[k], Y[k]]]); arr_N.set_UVC([0.5*NR[k]*nx], [0.5*NR[k]*ny]) # реакция N = mg√(1+f'²)
    info.set_text(f't = {T[k]:5.2f} c\nx = {X[k]:6.2f} м   y = {Y[k]:6.2f} м\n'
                  f'vx = {VX[k]:5.2f} м/с  vy = {VY[k]:5.2f} м/с\n|v| = {np.hypot(VX[k], VY[k]):5.2f} м/с   N/mg = {NR[k]:4.2f}')
    mark_a.set_data([T[k]], [Y[k]]); mark_b.set_data([T[k]], [VX[k]]); vline_b.set_xdata([T[k], T[k]])
    return point, trail, info

anim = FuncAnimation(fig, update, frames=n_frames, interval=50, blit=False)
anim.save('task_3_1_anim.gif', writer=PillowWriter(fps=20), dpi=70)         # GIF (воспроизводится везде)
display(Image(filename='task_3_1_anim.gif'))
plt.rcParams['animation.frame_format'] = 'jpeg'                       # кадры HTML-плеера в JPEG: файл ноутбука заметно меньше
display(HTML(anim.to_jshtml(fps=20)))                                         # HTML-плеер с кнопками паузы/воспроизведения
plt.close(fig)
```

## 5. Интерактивное управление (пауза / воспроизведение, ползунок $v_0$) через `matplotlib.widgets`
HTML-плеер выше уже содержит кнопки управления. Ниже — вариант на `matplotlib.widgets` (кнопка «Пауза/Пуск» и ползунок $v_0$, пересчитывающий кривую «на лету»). Для работы нужен интерактивный бэкенд: выполните в отдельной ячейке `%matplotlib widget` (нужен пакет `ipympl`) или `%matplotlib qt`, затем вызовите `ui = interactive_animation()`. Возвращаемый объект нужно **сохранить в переменной**, иначе сборщик мусора удалит анимацию.

```python
from matplotlib.widgets import Button, Slider

def interactive_animation():
    # Интерактивная анимация: Button — пауза/пуск, Slider — вертикальная скорость v0.
    fig, ax = plt.subplots(figsize=(10, 5))
    plt.subplots_adjust(bottom=0.25)                             # место под виджеты
    state = {'run': True, 't': 0.01, 'v0': 1.0}                  # состояние анимации
    line, = ax.plot([], [], 'b-'); point, = ax.plot([], [], 'ro', ms=10)
    txt = ax.text(0.02, 0.06, '', transform=ax.transAxes, family='monospace')
    ax.set_xlim(-0.5, 12); ax.set_ylim(-3, 0.5); ax.set_aspect('equal'); ax.grid(alpha=0.3)
    ax.set_xlabel('x, м'); ax.set_ylabel('y, м')

    def redraw_curve():
        xs = np.linspace(0, 12, 400); line.set_data(xs, curve_y(xs, 0, 0, state['v0']))

    def update(_):
        if state['run']:                                         # время идёт только если не на паузе
            state['t'] += 0.03
            if state['t'] > 2.5: state['t'] = 0.01               # по окончании — снова с начала
        x, y, vx, vy, Nr = motion(np.array([state['t']]), 0, 0, state['v0'], 1.0)
        point.set_data(x, y)
        txt.set_text(f"t = {state['t']:.2f} c  vx = {vx[0]:.2f}  vy = {vy[0]:.2f}  N/mg = {Nr[0]:.2f}")
        return point, txt

    ax_btn = plt.axes([0.15, 0.08, 0.15, 0.07]); btn = Button(ax_btn, 'Пауза / Пуск')
    ax_sl = plt.axes([0.45, 0.09, 0.40, 0.04]); sl = Slider(ax_sl, '$v_0$, м/с', 0.3, 3.0, valinit=1.0)
    def toggle(event): state['run'] = not state['run']           # переключаем паузу
    def change_v0(val): state['v0'] = val; state['t'] = 0.01; redraw_curve()   # новая v0 -> новая кривая
    btn.on_clicked(toggle); sl.on_changed(change_v0); redraw_curve()
    ani = FuncAnimation(fig, update, interval=40, cache_frame_data=False)
    return ani, btn, sl                                          # ссылки на виджеты обязательно сохранить!

# ui = interactive_animation()     # <- раскомментируйте после %matplotlib widget
```

## 6. Ответ

Искомая кривая — **полукубическая парабола (парабола Нейла)**:
$$
y=y_0-\frac32\Bigl(\frac{v_0^2}{3g}\Bigr)^{1/3}|x-x_0|^{2/3}\quad\Longleftrightarrow\quad (y_0-y)^3=\frac{9v_0^2}{8g}(x-x_0)^2,
$$
с остриём в точке $(x_0,y_0)$. Закон движения: $y=y_0-v_0t$, $x=x_0\pm\dfrac{\sqrt{8gv_0}}{3}t^{3/2}$. Реакция опоры $N=mg\sqrt{1+f'^2}=mg\sqrt{1+\dfrac{v_0}{2gt}}$.
